The data I choose is about EarlyON Child and Family Centres Locations. https://open.toronto.ca/dataset/earlyon-child-and-family-centres/

Centre Count by Ward: Illustrate the number of EarlyON Centres in each Ward to identify areas with higher or lower availability.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/Users/72xs/Desktop/dsi/visualization/02_activities/EarlyON Child and Family Centres Locations - geometry - 4326.csv')
#print(df.columns)
#print(df.head())
```

```python
ward_counts = df['ward_name'].value_counts().reset_index()
ward_counts.columns = ['Ward_Name', 'Centre Count']
#ward_counts
```

```python
plt.figure(figsize=(12, 8))
sns.barplot(data=ward_counts, x='Centre Count', y='Ward_Name', hue= 'Ward_Name',palette='viridis')
plt.title('Number of EarlyON Centres by Ward', fontsize=16)
plt.xlabel('Number of Centres', fontsize=14)
plt.ylabel('Ward Name', fontsize=16)
plt.xticks(rotation=45)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
